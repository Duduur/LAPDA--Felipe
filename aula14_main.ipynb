# Passo 8: Listando, Buscando e Filtrando

```python
from importlib import reload
import catalogo
from musica import Musica

# Recarrega a classe para usar as alteracoes salvas em catalogo.py.
Catalogo = reload(catalogo).Catalogo

m5 = Musica("Facil", "Alejandro Sanz", "Pop Latino", 343, 1997)
m6 = Musica ("Inner", "Elderbrook e Bob Moses", "Eletônica", 257, 2021)
m7 = Musica("Difícil", "Bruno & Marrone", "Sertanejo", 261, 1985)
m8 = Musica("teste", "Chitãozinho & Xororó", "Sertanejo", 293, 1990)
m9 = Musica("Coração partido", "Alejandro Sanz", "Pop Latino", 343, 1997)
m10 = Musica("Coraido", "o Sz", "Potino", 343, 2000)


musicas = [m5, m6, m7, m8, m9]

c2 = Catalogo()

for musica in musicas:
    c2.adicionar_musica(musica)
```

```python
print("Catalogo")
c2.listar_musicas()
```

```python
# buscar por nome
print("\n Buscar por nome: ")
for musica in c2.buscar_nome("inner"):
    print(musica)

# buscar por um cantor

print("Buscar por cantor: ")
for musica in c2.buscar_cantor("Alejandro"):
    print(musica)

# buscar por um genero

print("Buscar por genero: ")
for musica in c2.buscar_genero("Eletônica"):
    print(musica)

# buscar por um intervalo de duração

print("Buscar por intervalo de duração: ")
for musica in c2.buscar_duracao(2, 6):
    print(f"{musica.nome}")


# buscar por ano

print("Buscar por ano:")
for musica in c2.buscar_ano(1997):
    print(musica)
```

```python
removida = m10 in c2.musicas
if removida:
	c2.musicas.remove(m10)

print("Removida?", removida)

c2.listar_musicas()

print("A variavel m10 continua existindo", m10)
```
